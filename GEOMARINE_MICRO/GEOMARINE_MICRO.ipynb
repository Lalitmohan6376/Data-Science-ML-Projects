# **1. Latitude vs Microplastic Concentration**

Microplastic concentration varies across different latitude locations.

Some regions show noticeably higher concentrations.

There is no clear linear relationship between latitude and concentration.

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Latitude"],
    df["MP_conc__particles_cubic_metre_"]
)
plt.title("Latitude vs Microplastic Concentration")
plt.xlabel("Latitude")
plt.ylabel("Particles per Cubic Metre")
plt.grid(True)
plt.savefig("/content/Latitude vs Microplastic Concentration.png", dpi=300, bbox_inches="tight")
plt.show()
```

# **2. Longitude vs Microplastic Concentration**

Concentration changes across different longitude locations.

Some locations contain much higher values than others.

This indicates that microplastic concentration is spatially variable.

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Longitude"],
    df["MP_conc__particles_cubic_metre_"]
)
plt.title("Longitude vs Microplastic Concentration")
plt.xlabel("Longitude")
plt.ylabel("Particles per Cubic Metre")
plt.grid(True)
plt.savefig("/content/Longitude vs Microplastic Concentration.png", dpi=300, bbox_inches="tight")
plt.show()
```

# **3. Microplastic Concentration Over Time**

The concentration changes across different sampling dates.

Some dates show higher concentration values than others.

This indicates variation in microplastic concentration over the observed period.

```python
plt.figure(figsize=(9, 5))
plt.plot(
    df["Date"],
    df["MP_conc__particles_cubic_metre_"],
    marker="o"
)
plt.title("Microplastic Concentration Over Time")
plt.xlabel("Date")
plt.ylabel("Particles per Cubic Metre")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.savefig("/content/Microplastic Concentration Over Time.png", dpi=300, bbox_inches="tight")
plt.show()
```

# **4. Microplastic Concentration Distribution**

Most observations are concentrated within a particular value range.

A smaller number of observations have relatively high concentrations.

This shows that the dataset contains variation in microplastic levels.

```python
plt.figure(figsize=(8, 5))
plt.hist(
    df["MP_conc__particles_cubic_metre_"],
    bins=15
)
plt.title("Microplastic Concentration Distribution")
plt.xlabel("Particles per Cubic Metre")
plt.ylabel("Number of Observations")
plt.grid(True)
plt.savefig("/content/Microplastic Concentration Distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

# **5. Geographical Distribution of Sampling Locations**
The points show where the samples were collected geographically.

The sampling locations are spread across different latitude and longitude values.

This gives an overview of the spatial coverage of the dataset.

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Longitude"],
    df["Latitude"]
)
plt.title("Geographical Distribution of Sampling Locations")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.grid(True)
plt.savefig("/content/ Geographical Distribution of Sampling Locations.png", dpi=300, bbox_inches="tight")
plt.show()
```
